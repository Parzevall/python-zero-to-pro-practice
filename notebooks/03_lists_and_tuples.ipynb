Reimplement slicing: `my_slice(items, start=None, stop=None, step=None)` must return exactly what `items[start:stop:step]` returns, as a list, for **every** combination of inputs.

Reproduce the full contract:

- a missing `step` means `1`; `step == 0` raises `ValueError`;
- a missing `start`/`stop` means the beginning/end — which end depends on the sign of `step`;
- a negative bound has `len(items)` added to it;
- bounds are then **clamped** into range rather than rejected.

You may index with `items[i]` but you may not use slice syntax or the `slice` builtin.

<details><summary>Hint</summary>

Normalise `start` and `stop` into two plain integers first, then walk from one to the other by `step`. Every branch in that normalisation asks the same question: is `step` positive or negative? Note that a backwards walk has to be able to stop *before* index 0, which is why its floor is `-1` and not `0`.

</details>

<details><summary>Solution</summary>

```python
def my_slice(items, start=None, stop=None, step=None):
    n = len(items)
    if step is None:
        step = 1
    if step == 0:
        raise ValueError("slice step cannot be zero")
    if start is None:
        start = n - 1 if step < 0 else 0
    elif start < 0:
        start += n
        if start < 0:
            start = -1 if step < 0 else 0
    elif start >= n:
        start = n - 1 if step < 0 else n
    if stop is None:
        stop = -1 if step < 0 else n
    elif stop < 0:
        stop += n
        if stop < 0:
            stop = -1 if step < 0 else 0
    elif stop >= n:
        stop = n - 1 if step < 0 else n
    out = []
    i = start
    while i > stop if step < 0 else i < stop:
        out.append(items[i])
        i += step
    return out
```

**Why:** This is CPython's `PySlice_AdjustIndices` written out, and the asymmetry is the lesson: with a positive step the bounds clamp into `[0, n]`, but with a negative step they clamp into `[-1, n-1]`, because a backwards walk must be able to express 'stopped just past the front'. That single detail is why `items[::-1]` reaches element 0 while the naive `items[n-1:0:-1]` drops it, and why `items[-10::-1]` is empty rather than `[items[0]]`. Clamping instead of raising is what makes slicing total: no slice of any list is ever an error.
