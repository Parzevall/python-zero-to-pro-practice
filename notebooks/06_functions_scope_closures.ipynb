## Mini-project P-06 - Closure-built validation toolkit

Build a small validation library out of **factory functions** — each one returns a checker that closes over its configuration.

Write four factories. Every checker they return takes `(field, value)` and returns either `None` (fine) or a message string:

- `of_type(expected)` -> `f"{field}: expected {expected.__name__}, got {type(value).__name__}"`
- `in_range(lo, hi)` -> `f"{field}: {value} not in {lo}..{hi}"` when `value` is outside the inclusive range
- `non_empty()` -> `f"{field}: must not be empty"` when `value` is falsy
- `all_of(*checks)` -> runs the checkers in order and returns the **first** message, or `None` if they all pass

Then wire them into a module-level `SPEC` dict, in this order:

```python
SPEC = {
    "name": all_of(of_type(str), non_empty()),
    "age": all_of(of_type(int), in_range(13, 120)),
    "email": all_of(of_type(str), non_empty()),
}
```

Finally write `validate(record)` returning a **list of message strings** in `SPEC` order — at most one per field. A field missing from `record` gives `f"{field}: missing"` and its checkers are not run. A fully valid record gives `[]`.

<details><summary>Hint</summary>

Each factory's body is one `def` plus a `return` of that inner function — the configuration (`expected`, `lo`/`hi`, `checks`) is captured by the closure and never passed again. `all_of` is the interesting one: it closes over a tuple of other closures and stops at the first non-`None` result.

</details>

<details><summary>Solution</summary>

```python
def of_type(expected):
    def check(field, value):
        if not isinstance(value, expected):
            return f"{field}: expected {expected.__name__}, got {type(value).__name__}"
        return None

    return check


def in_range(lo, hi):
    def check(field, value):
        if not lo <= value <= hi:
            return f"{field}: {value} not in {lo}..{hi}"
        return None

    return check


def non_empty():
    def check(field, value):
        if not value:
            return f"{field}: must not be empty"
        return None

    return check


def all_of(*checks):
    def check(field, value):
        for one in checks:
            problem = one(field, value)
            if problem is not None:
                return problem
        return None

    return check


SPEC = {
    "name": all_of(of_type(str), non_empty()),
    "age": all_of(of_type(int), in_range(13, 120)),
    "email": all_of(of_type(str), non_empty()),
}


def validate(record):
    errors = []
    for field, check in SPEC.items():
        if field not in record:
            errors.append(f"{field}: missing")
            continue
        problem = check(field, record[field])
        if problem is not None:
            errors.append(problem)
    return errors
```

**Why:** Every rule here is one object that carries both its logic and its configuration, which is why `SPEC` can be plain data and `validate` knows nothing about types or ranges — adding a rule never touches the engine. The ordering inside `all_of` is load-bearing: the type check has to run first, because `in_range` would raise `TypeError` comparing a string to an int, and returning at the first failure is what keeps one bad field from producing a cascade of nonsense messages.
