<details><summary><b>Hint</b></summary>

`random.seed(7)` makes the sequence repeatable. `choice(seq)` picks one item; `sample(seq, k)` picks `k` items with no repeats.

</details>

<details><summary><b>Expected output</b></summary>

```text
pasta
['fries', 'greens']
```

</details>

<details><summary><b>Solution</b></summary>

```python

print(random.choice(menu))
print(random.sample(sides, 2))
```

`random` makes *pseudo*-random numbers from a starting value, the seed. Same seed, same sequence, so your output matches every time. Without a seed, Python seeds itself from the system and every run differs. The order of calls matters too: each call uses up part of the sequence.

</details>

```python
rng = random.Random(7)
print(rng.choice(menu))
print(rng.sample(sides, 2))
```

A private `random.Random(7)` gives the same sequence as `random.seed(7)`, but no other code can disturb it by calling `random` in between. Seniors use seeded generators in tests and simulations. For passwords and tokens they use `secrets` (NB31), because `random` is predictable by design.

</details>

### NB15-Q08 · L2 · itertools: groupby · Predict

`groupby` on data that isn't sorted. Predict the output.

**What does this print?**

```python
from itertools import groupby

words = ["apple", "avocado", "banana", "apricot", "blueberry"]
for letter, group in groupby(words, key=lambda w: w[0]):
    print(letter, list(group))
```

<details><summary><b>Hint</b></summary>

`groupby` starts a new group every time the key *changes*. It never looks back at earlier groups.

</details>

<details><summary><b>Expected output</b></summary>

```text
a ['apple', 'avocado']
b ['banana']
a ['apricot']
b ['blueberry']
```

</details>

<details><summary><b>Solution</b></summary>

`groupby` only groups items that sit **next to each other**. `apple` and `avocado` form one `a` group, then `banana` starts a `b` group, then `apricot` starts a *second* `a` group. So you get four groups, not two.

</details>

<details><summary><b>Senior dev solution</b></summary>

```python
from itertools import groupby

words = ["apple", "avocado", "banana", "apricot", "blueberry"]
for letter, group in groupby(sorted(words), key=lambda w: w[0]):
    print(letter, list(group))
```

Sort by the same key first, then group. Seniors reach for `groupby` when data already arrives sorted (log lines by time), and for `defaultdict(list)` (Q09) when it doesn't, since that needs no sort.

```python
def parse(line):
    day, clock, method, path, status, size = line.split()
    when = datetime.strptime(f"{day} {clock}", "%Y-%m-%d %H:%M:%S")
    return when, method, path, int(status), int(size)

records = sorted(parse(line) for line in log)
times = [rec[0] for rec in records]
statuses = Counter(rec[3] for rec in records)
paths = Counter(rec[2] for rec in records)
errors = sum(count for status, count in statuses.items() if status >= 400)

print("Requests:", len(records))
print("Span:", times[-1] - times[0])
print("Status codes:", ", ".join(f"{s} x{n}" for s, n in statuses.most_common()))
print(f"Error rate: {errors / len(records):.1%}")
print("Top paths:", ", ".join(f"{p} ({n})" for p, n in paths.most_common(3)))
print("Per hour:")
busiest_hour = None
busiest_count = 0
for hour, group in groupby(times, key=lambda t: t.hour):
    count = len(list(group))
    print(f"  {hour:02d}:00  {count}  {'#' * count}")
    if count > busiest_count:
        busiest_hour = hour
        busiest_count = count
print(f"Busiest hour: {busiest_hour:02d}:00 ({busiest_count} requests)")
print(f"Bytes served: {sum(rec[4] for rec in records):,}")
```

Parsing every line into real types first (a `datetime` and two `int`s) makes every later step easy: sorting puts the records in time order, which `groupby` needs and which also fixes the tie order in the status counts. Subtracting the first time from the last gives a `timedelta`, which prints as `H:MM:SS`. `.1%` multiplies by 100 and adds the `%` sign.

```python
from collections import namedtuple

Request = namedtuple("Request", "when method path status size")

def parse(line):
    day, clock, method, path, status, size = line.split()
    when = datetime.fromisoformat(f"{day} {clock}")
    return Request(when, method, path, int(status), int(size))

requests = sorted(map(parse, log))
statuses = Counter(r.status for r in requests)
paths = Counter(r.path for r in requests)
per_hour = Counter(r.when.hour for r in requests)
errors = sum(n for status, n in statuses.items() if status >= 400)

print("Requests:", len(requests))
print("Span:", requests[-1].when - requests[0].when)
print("Status codes:", ", ".join(f"{s} x{n}" for s, n in statuses.most_common()))
print(f"Error rate: {errors / len(requests):.1%}")
print("Top paths:", ", ".join(f"{p} ({n})" for p, n in paths.most_common(3)))
print("Per hour:")
for hour, count in per_hour.items():
    print(f"  {hour:02d}:00  {count}  {'#' * count}")
busiest_hour, busiest_count = per_hour.most_common(1)[0]
print(f"Busiest hour: {busiest_hour:02d}:00 ({busiest_count} requests)")
print(f"Bytes served: {sum(r.size for r in requests):,}")
```
