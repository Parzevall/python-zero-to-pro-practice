So far your data and the functions that work on it have lived apart: a
dict here, a function that takes the dict there. A *class* bundles them
together. You describe a kind of thing once (a bank account, a shape, a
parking spot), and then make as many *instances* of it as you need, each
with its own data.

This notebook covers the core: `__init__` and `self`, methods, class vs
instance attributes, `__str__`/`__repr__`/`__eq__`, inheritance with
`super()`, polymorphism, composition, privacy by convention, and looking
inside objects with `vars()`, `getattr()` and friends.

Properties, `@classmethod`, dataclasses, operator overloading and the rest
of the special methods come in NB21. Everything here works without them.

**One rule for output:** printing an object that has no `__str__` or
`__repr__` shows a memory address like `<__main__.Dog object at 0x7f...>`,
which changes every run. The questions always print attributes or define
`__str__`/`__repr__`, so your output can match exactly.

### NB14-Q01 · L1 · Classes and instances · Write

Write a class `Dog` whose `__init__` takes a `name` and stores it. Give it a
method `bark()` that **returns** the text `<name> says Woof!`. Create a dog
called `Rex` and print what `bark()` returns.

```text
Rex says Woof!
```

<details><summary><b>Hint</b></summary>

`def __init__(self, name):` runs when you write `Dog("Rex")`. Store the name with `self.name = name` so other methods can reach it through `self`.

</details>

<details><summary><b>Expected output</b></summary>

```text
Rex says Woof!
```

</details>

<details><summary><b>Solution</b></summary>

```python
class Dog:
    def __init__(self, name):
        self.name = name

    def bark(self):
        return self.name + " says Woof!"


rex = Dog("Rex")
print(rex.bark())
```

`class Dog:` describes what every dog has and does. `Dog("Rex")` builds one *instance* and calls `__init__` with `self` set to that new object, so `self.name = name` stores the name *on that dog*. `rex.bark()` calls the method with `self` set to `rex`.

</details>

<details><summary><b>Senior dev solution</b></summary>

```python
class Dog:
    def __init__(self, name):
        self.name = name

    def bark(self):
        return f"{self.name} says Woof!"


rex = Dog("Rex")
print(rex.bark())
```

An f-string (NB02) reads better than `+`. Seniors make methods *return* text rather than print it, so the caller decides whether to print it, log it, or put it on a web page.

Write a class `Product` with a `name` and a `price`, and a `__str__` method
so that printing a product shows its name and the price with 2 decimals.
Print `Product("Pen", 1.5)`, then use it inside an f-string.

```text
Pen: $1.50
In your basket: Pen: $1.50
```

<details><summary><b>Hint</b></summary>

`def __str__(self):` must **return** a string. `print()` and f-strings call it for you. `:.2f` formats a number with 2 decimals.

</details>

<details><summary><b>Expected output</b></summary>

```text
Pen: $1.50
In your basket: Pen: $1.50
```

</details>

<details><summary><b>Solution</b></summary>

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return f"{self.name}: ${self.price:.2f}"


pen = Product("Pen", 1.5)
print(pen)
print(f"In your basket: {pen}")
```

`__str__` is a *special method*: you never call it by name. `print(pen)`, `str(pen)` and `f"{pen}"` all call it to get the human-friendly text. Without it, you'd see a memory address.

</details>

<details><summary><b>Senior dev solution</b></summary>

```python

        return f"Product({self.name!r}, {self.price!r})"


pen = Product("Pen", 1.5)
print(pen)
print(f"In your basket: {pen}")
```

Seniors add a `__repr__` too (Q09): it's what you see in lists, the debugger and error messages, and the convention is that it looks like the code that would rebuild the object.

### NB14-Q05 · L1 · self · Error

The method is missing something. Which error does this raise?

**Which error does this raise?**

```python
class Counter:
    def __init__(self):
        self.count = 0

    def increment():
        self.count += 1


clicks = Counter()
clicks.increment()
```

<details><summary><b>Hint</b></summary>

`clicks.increment()` passes `clicks` to the method automatically. Where can it go?

</details>

<details><summary><b>Expected output</b></summary>

```text

```

</details>

<details><summary><b>Solution</b></summary>

`TypeError`: calling `clicks.increment()` is really `Counter.increment(clicks)`, so Python passes one argument, the instance. The method was defined with no parameters, so there's nowhere to put it ("takes 0 positional arguments but 1 was given"). Every method needs `self` as its first parameter.

</details>

```python

    def increment(self):
        self.count += 1

clicks.increment()
print(clicks.count)
```

`self` is a convention, not a keyword, but every Python developer expects it. Editors and linters flag a method without `self`, so seniors see this bug as a squiggle before they ever run the code.

</details>

### NB14-Q07 · L2 · Methods · Write

Write a `BankAccount` class with an `owner` and a `balance` (default `0`), and
two methods:

- `deposit(amount)` adds to the balance.
- `withdraw(amount)` takes money out, but if `amount` is more than the
  balance it prints `Declined: insufficient funds` and changes nothing.

Open an account for `"Ada"` with `100`, deposit `50`, withdraw `500`, withdraw
`30`, then print the balance.

```text
Declined: insufficient funds
Ada's balance: 120
```

<details><summary><b>Hint</b></summary>

`def __init__(self, owner, balance=0):`. Methods change the object through `self.balance`. An early `return` stops `withdraw` after declining.

</details>

<details><summary><b>Expected output</b></summary>

```text
Declined: insufficient funds
Ada's balance: 120
```

</details>

<details><summary><b>Solution</b></summary>

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            print("Declined: insufficient funds")
            return
        self.balance -= amount


account = BankAccount("Ada", 100)
account.deposit(50)
account.withdraw(500)
account.withdraw(30)
print(account.owner + "'s balance:", account.balance)
```

The object remembers its balance between calls, which is the whole point of a class: the data and the actions that change it live together. `100 + 50 = 150`, the `500` is declined, and `150 - 30 = 120`.

</details>

<details><summary><b>Senior dev solution</b></summary>


### NB14-Q09 · L2 · __repr__ · Write

Write a `Temperature` class holding `celsius`. Give it a `__str__` that shows
`21.5°C` and a `__repr__` that shows `Temperature(21.5)`. Print one
temperature on its own, then a list holding `Temperature(21.5)` and
`Temperature(-3)`.

```text
21.5°C
[Temperature(21.5), Temperature(-3)]
```

<details><summary><b>Hint</b></summary>

`print(obj)` uses `__str__`. Printing a *list* shows each item's `__repr__`.

</details>

<details><summary><b>Expected output</b></summary>

```text
21.5°C
[Temperature(21.5), Temperature(-3)]
```

</details>

<details><summary><b>Solution</b></summary>

```python
class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius

    def __str__(self):
        return str(self.celsius) + "°C"

    def __repr__(self):
        return "Temperature(" + str(self.celsius) + ")"


today = Temperature(21.5)
print(today)
print([today, Temperature(-3)])
```

`__str__` is for people (`21.5°C`). `__repr__` is for developers: it should say what the object *is*, ideally as code that would rebuild it. Containers like lists always show their items with `repr`, which is why the second line uses the other method.

</details>

<details><summary><b>Senior dev solution</b></summary>

```python
class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius

    def __str__(self):
        return f"{self.celsius}°C"

    def __repr__(self):
        return f"{type(self).__name__}({self.celsius!r})"


today = Temperature(21.5)
print(today)
print([today, Temperature(-3)])
```

`!r` (NB02) inserts the `repr` of the value, so a string would get its quotes. Using `type(self).__name__` instead of typing `Temperature` means a subclass shows its own name for free (Q31). If you write only one of the two, write `__repr__`: `str()` falls back to it.

### NB14-Q11 · L2 · Overriding methods · Write

Write a class `Animal` with a `name` and a `speak()` method that returns
`"..."`. Write `Cat` and `Cow` subclasses that override `speak()` to return
`"Meow"` and `"Moo"`. Loop over `Cat("Tom")`, `Cow("Daisy")` and
`Animal("Mystery")`, printing each name and sound.

```text
Tom: Meow
Daisy: Moo
Mystery: ...
```

<details><summary><b>Hint</b></summary>

`class Cat(Animal):` inherits everything from `Animal`, including `__init__`. Defining `speak` again in `Cat` replaces the inherited one.

</details>

<details><summary><b>Expected output</b></summary>

```text
Tom: Meow
Daisy: Moo
Mystery: ...
```

</details>

<details><summary><b>Solution</b></summary>

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return "..."


class Cat(Animal):
    def speak(self):
        return "Meow"


class Cow(Animal):
    def speak(self):
        return "Moo"


for animal in [Cat("Tom"), Cow("Daisy"), Animal("Mystery")]:
    print(animal.name + ":", animal.speak())
```

`Cat` and `Cow` don't define `__init__`, so they inherit `Animal`'s and get a `name` for free. Each overrides `speak`. The loop calls `animal.speak()` without caring what kind of animal it is; Python finds the right method for each object. That is *polymorphism*.

</details>


When subclasses differ only by a value, seniors override a class attribute instead of a whole method. Overriding methods is for different *behaviour*; overriding data is for different *settings*.

Its `withdraw` checks the minimum, then `super().withdraw(amount)` does the real work, including the history entry, so that logic exists in one place. Each account makes its own `history` list in `__init__`; a class-level list would mix both accounts' entries (Q18). Raising an exception means a refused withdrawal can't be silently ignored; the loop decides how to report it. A savings account *is a* bank account, but not the reverse.

</details>

```python
class InsufficientFunds(Exception):
    pass


class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
        self.history = []

    def deposit(self, amount):
        self.balance += amount
        self.history.append(f"+{amount}")

    def withdraw(self, amount):
        if amount > self.balance:
            raise InsufficientFunds(f"{self.owner} cannot withdraw {amount}")
        self.balance -= amount
        self.history.append(f"-{amount}")

    def __str__(self):
        return f"{self.owner}: {self.balance} {self.history}"


class SavingsAccount(BankAccount):
    def __init__(self, owner, balance=0, minimum=100):
        super().__init__(owner, balance)
        self.minimum = minimum

    def withdraw(self, amount):
        if self.balance - amount < self.minimum:
            raise InsufficientFunds(f"{self.owner} must keep {self.minimum}")
        super().withdraw(amount)


current = BankAccount("Ada", 50)
savings = SavingsAccount("Ada savings", 300)
steps = [(current.withdraw, 80), (savings.withdraw, 150), (savings.withdraw, 100),
         (current.deposit, 40), (current.withdraw, 80)]

for operation, amount in steps:
    try:
        operation(amount)
    except InsufficientFunds as e:
        print(f"Refused: {e}")

print(current)
print(savings)
print(isinstance(savings, BankAccount), issubclass(BankAccount, SavingsAccount))
```

Storing bound methods (`current.withdraw`) in the steps removes the `getattr` and the string names entirely, so a typo is caught the moment the list is built. Seniors also reject negative amounts and keep money as whole cents or `Decimal` (NB16).

```python
class ParkingError(Exception):
    pass


class NoSpaceError(ParkingError):
    pass


class NotParkedError(ParkingError):
    pass


class Vehicle:
    size = 0

    def __init__(self, plate):
        self.plate = plate

    def __str__(self):
        return type(self).__name__ + " " + self.plate


class Motorbike(Vehicle):
    size = 1


class Car(Vehicle):
    size = 2


class Van(Vehicle):
    size = 3


class Spot:
    def __init__(self, spot_id, size):
        self.spot_id = spot_id
        self.size = size
        self.vehicle = None

    def fits(self, vehicle):
        return self.vehicle is None and vehicle.size <= self.size


class Level:
    def __init__(self, name, sizes):
        self.name = name
        self.spots = []
        number = 1
        for size in sizes:
            self.spots.append(Spot(name + str(number), size))
            number += 1


class ParkingLot:
    def __init__(self, levels):
        self.levels = levels

    def park(self, vehicle):
        for level in self.levels:
            for spot in level.spots:
                if spot.fits(vehicle):
                    spot.vehicle = vehicle
                    return spot.spot_id
        raise NoSpaceError(str(vehicle))

    def leave(self, plate):
        for level in self.levels:
            for spot in level.spots:
                if spot.vehicle is not None and spot.vehicle.plate == plate:
                    spot.vehicle = None
                    return spot.spot_id
        raise NotParkedError(plate)

    def status(self):
        lines = []
        total_free = 0
        for level in self.levels:
            free = 0
            spot_lines = []
            for spot in level.spots:
                if spot.vehicle is None:
                    free += 1
                    spot_lines.append("  " + spot.spot_id + ": free")
                else:
                    spot_lines.append("  " + spot.spot_id + ": " + str(spot.vehicle))
            total_free += free
            lines.append(f"Level {level.name}: {free} of {len(level.spots)} free")
            lines.extend(spot_lines)
        lines.append("Total free: " + str(total_free))
        return lines


def park(lot, vehicle):
    try:
        print("Parked", vehicle, "at", lot.park(vehicle))
    except ParkingError:
        print("Could not park", vehicle)

    try:
        print(plate, "left", lot.leave(plate))
    except ParkingError:
```

Each class has one job. A `Vehicle` knows its size, a `Spot` knows whether something fits, a `Level` builds its spots, and the `ParkingLot` searches them. The car park *has* levels, which *have* spots, which *have* a vehicle: composition all the way down, with inheritance only where something really *is a* kind of something else.

The traps: the first car can't use `A1` (too small), the second van doesn't fit anywhere because `B1` is the only size-3 spot, and the new motorbike takes the freed `A2` because it's the first spot that fits. Catching the base `ParkingError` handles both failure types at once.

```python
class ParkingError(Exception):
    pass


class NoSpaceError(ParkingError):
    pass


class NotParkedError(ParkingError):
    pass


class Vehicle:
    size = 0

    def __init__(self, plate):
        self.plate = plate

    def __str__(self):
        return f"{type(self).__name__} {self.plate}"


class Motorbike(Vehicle):
    size = 1


class Car(Vehicle):
    size = 2


class Van(Vehicle):
    size = 3


class Spot:
    def __init__(self, spot_id, size):
        self.spot_id = spot_id
        self.size = size
        self.vehicle = None

    def fits(self, vehicle):
        return self.vehicle is None and vehicle.size <= self.size

    def __str__(self):
        return f"  {self.spot_id}: {self.vehicle or 'free'}"


class Level:
    def __init__(self, name, sizes):
        self.name = name
        self.spots = [Spot(f"{name}{number}", size) for number, size in enumerate(sizes, 1)]

    def free_count(self):
        return sum(spot.vehicle is None for spot in self.spots)


class ParkingLot:
    def __init__(self, levels):
        self.levels = levels
        self._where = {}

    def _all_spots(self):
        return [spot for level in self.levels for spot in level.spots]

    def park(self, vehicle):
        if vehicle.plate in self._where:
            raise ParkingError(f"{vehicle.plate} is already parked")
        for spot in self._all_spots():
            if spot.fits(vehicle):
                spot.vehicle = vehicle
                self._where[vehicle.plate] = spot
                return spot.spot_id
        raise NoSpaceError(str(vehicle))

    def leave(self, plate):
        spot = self._where.pop(plate, None)
        if spot is None:
            raise NotParkedError(plate)
        spot.vehicle = None
        return spot.spot_id

    def status(self):
        lines = []
        for level in self.levels:
            lines.append(f"Level {level.name}: {level.free_count()} of {len(level.spots)} free")
            lines.extend(str(spot) for spot in level.spots)
        lines.append(f"Total free: {sum(level.free_count() for level in self.levels)}")
        return lines


lot = ParkingLot([Level("A", [1, 2, 2]), Level("B", [3, 2])])
steps = [("park", Car("KT-001")), ("park", Motorbike("MB-777")), ("park", Van("VN-900")),
         ("park", Car("KT-002")), ("park", Van("VN-901")), ("leave", "KT-001"),
         ("leave", "ZZ-999"), ("park", Motorbike("MB-778"))]
for action, target in steps:
    try:
        if action == "park":
            print(f"Parked {target} at {lot.park(target)}")
        else:
            print(f"{target} left {lot.leave(target)}")
    except ParkingError:
        print(f"Could not {'park' if action == 'park' else 'remove'} {target}")
print("\n".join(lot.status()))
```

The senior lot keeps a `_where` dict from plate to spot, so `leave` is one lookup instead of a search, and parking the same car twice is refused. Each object can describe itself (`Spot.__str__`, `Level.free_count`), so `status` only assembles lines. In NB21, `Vehicle` could become an abstract base class and the vehicle sizes an `Enum`.
