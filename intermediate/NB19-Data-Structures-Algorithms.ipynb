**Big-O in one paragraph.** Big-O describes how the work grows as the
input grows. O(1) means "the same work however big the data is" (a dict
lookup). O(n) means "one pass over the data" (`x in some_list`). O(n²)
means "a pass for every item" (two nested loops). O(log n) means "halve the
problem each step" (binary search). You won't time anything here: you'll
count steps instead, because counts are the same on every computer.

```python
n = 16

linear = 0
```

Find the two items whose prices add up exactly to `budget` (there's exactly
one such pair). Print their positions in the list, smaller first.

Loop **once**: keep a dict that maps each price you've already seen to its
index.

```text
3 4
```

**Starting code (already in the cell below):**

```python
prices = [12, 7, 30, 18, 5]
budget = 23
```

<details><summary><b>Hint</b></summary>

For each price, the partner you need is `budget - price`. Have you already seen it? A dict answers that in one step.

</details>

<details><summary><b>Expected output</b></summary>

```text
3 4
```

</details>

<details><summary><b>Solution</b></summary>

```python
seen = {}
for i, price in enumerate(prices):
    needed = budget - price
    if needed in seen:
        print(seen[needed], i)
        break
    seen[price] = i
```

The obvious way checks every pair with two loops: O(n²). Here each price asks the dict "have I met my partner?", and a dict lookup is O(1) on average, so the whole thing is O(n). Checking *before* storing stops a price from pairing with itself.

</details>

<details><summary><b>Senior dev solution</b></summary>

```python
def pair_for(prices, budget):
    index_of = {}
    for i, price in enumerate(prices):
        if budget - price in index_of:
            return index_of[budget - price], i
        index_of[price] = i
    return None

print(*pair_for(prices, budget))
```

A function returns the answer instead of printing from inside the loop, and says clearly what happens when there's no pair (`None`). This is the famous "two-sum" interview question; trading memory (the dict) for speed is the whole trick.

</details>

```python
window = sum(readings[:k])
best = window
best_start = 0
for i in range(k, len(readings)):
    window += readings[i] - readings[i - k]
    if window > best:
        best = window
        best_start = i - k + 1
```

### NB19-Q13 · L2 · Insertion sort · Fix

This insertion sort should print `[1, 2, 5, 7, 9]`, but one item ends up in
the wrong place. Fix it.

**Buggy code (copied into the cell below):**

```python
def insertion_sort(items):
    result = items[:]
    for i in range(1, len(result)):
        current = result[i]
        j = i - 1
        while j > 0 and result[j] > current:
            result[j + 1] = result[j]
            j -= 1
        result[j + 1] = current

print(insertion_sort([5, 2, 9, 1, 7]))
```

<details><summary><b>Hint</b></summary>

Which index does the `while` loop never compare `current` with?

</details>

<details><summary><b>Expected output</b></summary>

```text
[1, 2, 5, 7, 9]
```

</details>

<details><summary><b>Solution</b></summary>

```python
def insertion_sort(items):
    result = items[:]
    for i in range(1, len(result)):
        current = result[i]
        j = i - 1
        while j >= 0 and result[j] > current:
            result[j + 1] = result[j]
            j -= 1
        result[j + 1] = current
    return result

print(insertion_sort([5, 2, 9, 1, 7]))
```

Insertion sort takes each item and shifts bigger items right until it finds its place. With `j > 0` the loop stops *before* index 0, so nothing is ever compared with the first item and `5` never moves. `j >= 0` lets the shifting reach the front. It's O(n²) in general, but nearly O(n) on data that's already almost sorted.

</details>

<details><summary><b>Senior dev solution</b></summary>

```python
print(sorted([5, 2, 9, 1, 7]))
```

In real code, `sorted()`. Python's Timsort is O(n log n), stable, and uses insertion sort itself on short runs, because insertion sort is very fast on small or nearly sorted data. Writing it yourself is for understanding, not for production.

</details>

````python
def insertion_sort(items):
    result = items[:]
    for i in range(1, len(result)):
        current = result[i]
        j = i - 1
        while j > 0 and result[j] > current:
            result[j + 1] = result[j]
            j -= 1
        result[j + 1] = current
    return result

print(insertion_sort([5, 2, 9, 1, 7]))
````

```python
class Node:
    def __init__(self, value, next=None):
        self.value = value
        self.next = next


class LinkedList:
    def __init__(self):
        self.head = None

    def push_front(self, value):
        self.head = Node(value, self.head)

    def append(self, value):
        if self.head is None:
            self.head = Node(value)
            return
        node = self.head
        while node.next is not None:
            node = node.next
        node.next = Node(value)

    def to_list(self):
        values = []
        node = self.head
        while node is not None:
            values.append(node.value)
            node = node.next
        return values

numbers.append(3)
numbers.push_front(1)
numbers.append(4)
print(numbers.to_list())
print("head:", numbers.head.value)
```

<details><summary><b>Hint</b></summary>

In `merge`, compare the front items of both lists and take the smaller one. When one list runs out, add the rest of the other. A list of 0 or 1 items is already sorted: that's your base case.

</details>

<details><summary><b>Expected output</b></summary>

```text
[1, 2, 3, 4, 9, 10]
[3, 9, 10, 27, 38, 43, 82]
```

</details>

<details><summary><b>Solution</b></summary>

```python
def merge(left, right):
    result = []
    i = 0
    j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result


def merge_sort(items):
    if len(items) <= 1:
        return items
    middle = len(items) // 2
    return merge(merge_sort(items[:middle]), merge_sort(items[middle:]))


print(merge([1, 4, 9], [2, 3, 10]))
print(merge_sort([38, 27, 43, 3, 9, 82, 10]))
```

Divide and conquer: halving gives about log n levels, and merging each level touches every item once, so merge sort is O(n log n) every time, even in the worst case. Using `<=` (not `<`) takes from the left list on ties, which keeps equal items in their original order: the sort is **stable**.

```python
            size += 1
```

```python
                    land.remove(step)
```

<details><summary><b>Hint</b></summary>

Make a table `dp` with one more row than `a` has letters and one more column than `b`. `dp[i][j]` is the answer for the first `i` letters of `a` and the first `j` of `b`. Build the rows with a comprehension, **not** `[[0] * cols] * rows` (see Q39). Then walk back from the bottom-right corner to collect the letters.

</details>

<details><summary><b>Expected output</b></summary>

```text
AGGTAB / GXTXAYB: 4 'GTAB'
ABC / DEF: 0 ''
```

</details>

<details><summary><b>Solution</b></summary>

```python
for a, b in pairs:
    dp = [[0] * (len(b) + 1) for row in range(len(a) + 1)]
    for i in range(1, len(a) + 1):
        for j in range(1, len(b) + 1):
            if a[i - 1] == b[j - 1]:
                dp[i][j] = dp[i - 1][j - 1] + 1
            else:
                dp[i][j] = max(dp[i - 1][j], dp[i][j - 1])

    letters = []
    i = len(a)
    j = len(b)
    while i > 0 and j > 0:
        if a[i - 1] == b[j - 1]:
            letters.append(a[i - 1])
            i -= 1
            j -= 1
        elif dp[i - 1][j] >= dp[i][j - 1]:
            i -= 1
        else:
            j -= 1
    common = "".join(reversed(letters))
    print(f"{a} / {b}: {dp[len(a)][len(b)]} {common!r}")
```

If the last letters match, they're part of the answer: take `1 +` the answer without both. If not, drop the last letter of one string or the other and keep the better result. Every cell uses cells already filled, so one pass over the table works: O(len(a) × len(b)). Walking back follows the same choices in reverse, collecting letters where they matched.

```python
from functools import cache

def lcs(a, b):
    @cache
    def best(i, j):
        if i == len(a) or j == len(b):
            return ""
        if a[i] == b[j]:
            return a[i] + best(i + 1, j + 1)
        return max(best(i + 1, j), best(i, j + 1), key=len)
    return best(0, 0)

for a, b in pairs:
    common = lcs(a, b)
    print(f"{a} / {b}: {len(common)} {common!r}")
```

The top-down version with `@cache` (NB11) reads like the definition, and the cache turns it into the same O(n × m) work. For long inputs the bottom-up table wins: no recursion limit, and less memory per cell. `difflib` (NB15) uses ideas like this to compare whole files.

### NB19-Q37 · L5 · Sliding window + hashing · Write

For each word, find the longest stretch of letters with **no repeated
letter**. Print its length and the stretch (the first one, if there's a
tie).

```text
'abcabcbb': 3 'abc'
'bbbbb': 1 'b'
'pwwkew': 3 'wke'
'abba': 2 'ab'
'': 0 ''
```

**Starting code (already in the cell below):**

```python
words = ["abcabcbb", "bbbbb", "pwwkew", "abba", ""]
```

<details><summary><b>Hint</b></summary>

Slide a window `left..right`. Keep a dict of the last index where each letter appeared. When the new letter was last seen **inside** the window, move `left` just past it. `abba` checks that `left` never moves backwards.

</details>

<details><summary><b>Expected output</b></summary>

```text
'abcabcbb': 3 'abc'
'bbbbb': 1 'b'
'pwwkew': 3 'wke'
'abba': 2 'ab'
'': 0 ''
```

</details>

<details><summary><b>Solution</b></summary>

```python
for word in words:
    last_seen = {}
    left = 0
    best_start = 0
    best_length = 0
    for right, letter in enumerate(word):
        if letter in last_seen and last_seen[letter] >= left:
            left = last_seen[letter] + 1
        last_seen[letter] = right
        if right - left + 1 > best_length:
            best_length = right - left + 1
            best_start = left
    print(f"{word!r}: {best_length} {word[best_start:best_start + best_length]!r}")
```

Each letter enters the window once and `left` only ever moves forward, so it's O(n). The trap is `abba`: at the last `a`, `last_seen['a']` is `0`, but `left` is already `2` (after the second `b`). Jumping to `0 + 1` would move the window *backwards* and count `bba`, which repeats `b`. Checking `last_seen[letter] >= left` ignores sightings from before the window.

</details>

<details><summary><b>Senior dev solution</b></summary>

```python
def longest_unique_run(word):
    last_seen = {}
    left = best_start = best_length = 0
    for right, letter in enumerate(word):
        left = max(left, last_seen.get(letter, -1) + 1)
        last_seen[letter] = right
        if right - left + 1 > best_length:
            best_start, best_length = left, right - left + 1
    return word[best_start:best_start + best_length]

for word in words:
    run = longest_unique_run(word)
    print(f"{word!r}: {len(run)} {run!r}")
```

`max(left, ...)` states the rule "the window never moves backwards" in one expression, and `.get(letter, -1)` removes the membership check. Returning the stretch itself (not two numbers) means the caller gets the length for free.

</details>

### NB19-Q38 · L5 · Dijkstra · Write

`roads` gives one-way road lengths in km. Print the shortest distance from
`A` to every place, in alphabetical order. A place you can't reach prints
`unreachable`.

```text
A: 0
B: 3
C: 1
D: 4
E: unreachable
```

Watch out: the direct road `A → B` is 4 km, but `A → C → B` is shorter.

**Starting code (already in the cell below):**

```python
import heapq

roads = {
    "A": {"B": 4, "C": 1},
    "B": {"D": 1},
    "C": {"B": 2, "D": 7},
    "D": {},
    "E": {"A": 2},
}
```

<details><summary><b>Hint</b></summary>

Keep a dict `distance` (start: `{"A": 0}`) and a heap of `(distance, place)`. Always expand the closest place not yet finished. An entry popped with a bigger distance than the one you've since found is out of date: skip it.

</details>

<details><summary><b>Expected output</b></summary>

```text
A: 0
B: 3
C: 1
D: 4
E: unreachable
```

</details>

<details><summary><b>Solution</b></summary>

```python
distance = {"A": 0}
heap = [(0, "A")]
while heap:
    dist, place = heapq.heappop(heap)
    if dist > distance[place]:
        continue
    for neighbour, length in roads[place].items():
        new_dist = dist + length
        if neighbour not in distance or new_dist < distance[neighbour]:
            distance[neighbour] = new_dist
            heapq.heappush(heap, (new_dist, neighbour))

for place in sorted(roads):
    if place in distance:
        print(f"{place}: {distance[place]}")
    else:
        print(f"{place}: unreachable")
```

Dijkstra always finishes the closest unfinished place next. Since roads can't have negative lengths, nothing found later can beat it. `B` is first pushed at 4, then improved to 3 via `C` and pushed again; the stale `(4, "B")` entry is skipped when it's popped. `E` has a road *to* `A`, but no road leads to `E`, so it's unreachable. With a heap, the whole run is O((places + roads) log places).

</details>

<details><summary><b>Senior dev solution</b></summary>

```python
import math

def shortest_distances(graph, start):
    distance = {start: 0}
    heap = [(0, start)]
    while heap:
        dist, place = heapq.heappop(heap)
        if dist > distance[place]:
            continue
        for neighbour, length in graph[place].items():
            new_dist = dist + length
            if new_dist < distance.get(neighbour, math.inf):
                distance[neighbour] = new_dist
                heapq.heappush(heap, (new_dist, neighbour))
    return distance

distance = shortest_distances(roads, "A")
for place in sorted(roads):
    print(f"{place}: {distance.get(place, 'unreachable')}")
```

`distance.get(neighbour, math.inf)` treats "never seen" as infinitely far, which removes the `not in` check. Putting the algorithm in a function with the graph and start as parameters makes it reusable, as the mini-project will show. Dijkstra breaks on negative edges; that needs Bellman-Ford.

</details>

<details><summary><b>Solution</b></summary>

```python
def fastest_route(start, goal):
    minutes = {start: 0}
    came_from = {start: None}
    heap = [(0, start)]
    while heap:
        time_so_far, place = heapq.heappop(heap)
        if time_so_far > minutes[place]:
            continue
        if place == goal:
            route = []
            while place is not None:
                route.append(place)
                place = came_from[place]
            route.reverse()
            return time_so_far, route
        for neighbour, drive in city[place].items():
            new_time = time_so_far + drive
            if neighbour not in minutes or new_time < minutes[neighbour]:
                minutes[neighbour] = new_time
                came_from[neighbour] = place
                heapq.heappush(heap, (new_time, neighbour))
    return None


for start, goal in trips:
    result = fastest_route(start, goal)
    if result is None:
        print(f"{start} -> {goal}: no route")
        continue
    total, route = result
    print(f"{start} -> {goal}: {total} min")
    if len(route) == 1:
        print("  (you are already there)")
    for step, (here, there) in enumerate(zip(route, route[1:]), start=1):
        print(f"  {step}. {here} -> {there} ({city[here][there]} min)")
```

Dijkstra pops places in order of travel time, so the moment the goal comes off the heap its time is final and you can stop early. `came_from` is updated together with `minutes`, so it always points back along the fastest way found so far. The direct `Station → Museum` road (18 min) loses to the 14-minute route through `Market` and `Park`, and `Harbour → Airport` goes back through `Station` rather than taking the 25-minute road. `zip(route, route[1:])` pairs each place with the next one, which turns a list of places into a list of roads.

```python
import math
from itertools import pairwise

def fastest_route(graph, start, goal):
    best = {start: 0}
    came_from = {start: None}
    heap = [(0, start)]
    while heap:
        elapsed, place = heapq.heappop(heap)
        if elapsed > best[place]:
            continue
        if place == goal:
            route = [place]
            while (place := came_from[place]) is not None:
                route.append(place)
            return elapsed, route[::-1]
        for neighbour, drive in graph[place].items():
            if elapsed + drive < best.get(neighbour, math.inf):
                best[neighbour] = elapsed + drive
                came_from[neighbour] = place
                heapq.heappush(heap, (elapsed + drive, neighbour))
    return None

    total, route = result
    lines = [f"{start} -> {goal}: {total} min"]
    if len(route) == 1:
        lines.append("  (you are already there)")
    for step, (here, there) in enumerate(pairwise(route), start=1):
        lines.append(f"  {step}. {here} -> {there} ({graph[here][there]} min)")
    return lines


for start, goal in trips:
    print("\n".join(describe(city, start, goal)))
```
