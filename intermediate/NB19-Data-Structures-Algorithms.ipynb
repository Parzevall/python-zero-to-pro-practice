Write `quick_sort(items)` recursively: a list of 0 or 1 items is already

**Expected output:**


<details><summary><b>Solution</b></summary>

```python
def quick_sort(items):
    if len(items) <= 1:
        return items
    pivot = items[0]
    smaller = [x for x in items[1:] if x < pivot]
    larger = [x for x in items[1:] if x >= pivot]
    return quick_sort(smaller) + [pivot] + quick_sort(larger)
print(quick_sort(times))
```

```python
def insertion_sort(items):
    for i in range(1, len(items)):
        value = items[i]
        j = i - 1
        while j >= 0 and items[j] > value:
            items[j + 1] = items[j]
            j -= 1
        items[j + 1] = value
```

```python
def merge_sort(items):
    if len(items) <= 1:
        return items
    mid = len(items) // 2
    left = merge_sort(items[:mid])
    right = merge_sort(items[mid:])
    merged = []
    i = 0
    j = 0
    while i < len(left) and j < len(right):

            i += 1
```

### Q17 · Real-life

The `flights` dict gives the ticket price of each direct flight (it works
both ways). Write `cheapest(flights, start)` using Dijkstra's algorithm
with a heap: it returns a dict of the lowest total price from `start` to

**Expected output:**

<details><summary><b>Solution</b></summary>

```python
import heapq
def cheapest(flights, start):
    best = {start: 0}
    heap = [(0, start)]
    while heap:
        price, city = heapq.heappop(heap)
        if price > best[city]:
            continue
        for nxt, cost in flights[city].items():
            total = price + cost
            if total < best.get(nxt, float("inf")):
                best[nxt] = total
                heapq.heappush(heap, (total, nxt))
    return best
prices = cheapest(flights, "Oslo")
for city in sorted(prices):
    print(city, prices[city])
```

```python
def fewest_coins(coins, amount):
    best = [0] + [float("inf")] * amount
    for a in range(1, amount + 1):
        for coin in coins:
            if coin <= a and best[a - coin] + 1 < best[a]:
                best[a] = best[a - coin] + 1
```

not necessarily next to each other), using a table of

**Expected output:**

```text
4
```

<details><summary><b>Solution</b></summary>

```python
def lcs_length(a, b):
    table = [[0] * (len(b) + 1) for _ in range(len(a) + 1)]
    for i in range(1, len(a) + 1):
        for j in range(1, len(b) + 1):
            if a[i - 1] == b[j - 1]:
                table[i][j] = table[i - 1][j - 1] + 1
            else:
                table[i][j] = max(table[i - 1][j], table[i][j - 1])
    return table[len(a)][len(b)]
```

```python
from collections import deque
def fewest_roads(roads, start, goal):
    queue = deque([[start]])
    visited = {start}
    while queue:
        route = queue.popleft()
        if route[-1] == goal:
            return route
        for nxt in sorted(roads[route[-1]]):
            if nxt not in visited:
                visited.add(nxt)
                queue.append(route + [nxt])
```

```python
def quickest_route(roads, start, goal):
    best = {start: 0}
    came_from = {}

        if minutes > best[place]:

        for nxt, cost in roads[place].items():

    route.reverse()
    return best[goal], route
```
