```python
    cache = {}
```

<details><summary><b>Solution</b></summary>

```python
def once(func):
    done = False
    result = None
    def wrapper(*args, **kwargs):
        nonlocal done, result
        if not done:
            result = func(*args, **kwargs)
            done = True
        return result
    return wrapper
connect_once = once(connect)
for host in ["a.com", "b.com", "c.com"]:
    print(connect_once(host))
```
