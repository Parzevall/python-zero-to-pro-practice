# NB09 · Functions Advanced

`intermediate/` · notebook 9 of 32 · 41 questions + a mini-project

In NB07 you wrote functions with a fixed list of parameters. This notebook
makes functions flexible and powerful: they can take any number of
arguments, insist on keywords, be stored in dicts, be passed around like
any other value, and even remember things between calls.

Three ideas carry the whole notebook:

- **`*args` and `**kwargs`** collect extra arguments into a tuple and a
  dict. In a *call*, `*` and `**` do the opposite: they spread a list or
  dict out into separate arguments.
- **Functions are values.** `shout` (no brackets) is the function itself;
  `shout()` runs it. You can put the first one in a variable, a list or a
  dict, and call it later.
- **Closures.** A function defined inside another function can still see
  the outer function's variables after the outer one has returned.
  `nonlocal` lets it change them.

Comprehensions and `key=` sorting come in NB10, and decorators (the `@`
syntax) in NB23, so these questions use plain loops and wrap functions by
hand.

Write a function `total(*nums)` that returns the sum of however many

</details>

<details><summary><b>Expected output</b></summary>

### NB09-Q10 · L2 · Higher-order functions · Write

Write `apply_to_all(func, values)` that returns a **new list** made by
calling `func` on each value. Print `apply_to_all(abs, [-3, 1, -2])` and

```text
[3, 1, 2]
['TEA', 'JAM']
```

<details><summary><b>Hint</b></summary>

Loop over `values`, call `func(value)` and append the result to a new list.

</details>

<details><summary><b>Expected output</b></summary>

```text
[3, 1, 2]
['TEA', 'JAM']
```

</details>

<details><summary><b>Solution</b></summary>

```python
def apply_to_all(func, values):
    results = []
    for value in values:
        results.append(func(value))
    return results


print(apply_to_all(abs, [-3, 1, -2]))
print(apply_to_all(str.upper, ["tea", "jam"]))
```

A *higher-order function* takes a function as an argument. `apply_to_all` doesn't know or care what `func` does; it calls it on each item. `str.upper` is the `upper` method reached through the `str` type, so `str.upper("tea")` is the same as `"tea".upper()`.

</details>

<details><summary><b>Senior dev solution</b></summary>

```python
print(list(map(abs, [-3, 1, -2])))
print([word.upper() for word in ["tea", "jam"]])
```

You have written your own version of the built-in `map()`. NB10 covers `map` and list comprehensions, which seniors reach for instead of writing this helper.

`run_later` should receive a **function** and call it. This crashes. Fix it

**Buggy code (copied into the cell below):**

```python
def report():

def run_later(task):
```

```python
    def on(event, handler):
        handlers.setdefault(event, []).append(handler)

            called += 1
        return called
```

```python
def make_event_bus():
    """Return (on, off, emit, listeners) functions sharing one handler registry."""
    handlers = {}

    def on(event, handler):
        handlers.setdefault(event, []).append(handler)
        return handler

    def off(event, handler):
        registered = handlers.get(event, [])
        if handler in registered:
            registered.remove(handler)

    def emit(event, *args, **kwargs):
        current = list(handlers.get(event, []))
        for handler in current:
            handler(*args, **kwargs)
        return len(current)

    def listeners(event):
        return [handler.__name__ for handler in handlers.get(event, [])]

    return on, off, emit, listeners
```
