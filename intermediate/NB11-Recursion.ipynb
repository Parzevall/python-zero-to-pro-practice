# NB11 · Recursion

`intermediate/` · notebook 11 of 32 · 31 questions + a mini-project

A recursive function is a function that calls itself. Each call solves a
smaller copy of the same problem, until the problem is so small the answer
is obvious. That tiny problem is the **base case**; the step that shrinks
the problem is the **recursive case**.

Every call that hasn't finished yet waits on the **call stack**. Python
keeps the stack fairly short (about 1000 calls by default), so a function
that never reaches its base case stops with a `RecursionError`.

Recursion is the natural tool for things that contain smaller copies of
themselves: nested lists, folders inside folders, comment threads, and
"try every option" puzzles. For a straight line of work, a loop is usually
simpler. This notebook shows you both, and when to pick which.

You can use everything up to NB10: functions, closures, comprehensions
and sorting. `raise` and `class` come in NB12 and NB14, so the solutions
here don't use them.

### NB11-Q01 · L1 · Base Case · Write

Write a recursive function `countdown(n)` that prints the numbers from `n`
down to `1`, one per line, then prints `Liftoff!`. Call it with `3`.

```text
3
2
1
Liftoff!
```

Use no loop: the function should call itself.

<details><summary><b>Hint</b></summary>

When `n` reaches `0` there's nothing left to count: print `Liftoff!` and stop. Otherwise print `n` and let a call with `n - 1` do the rest.

</details>

<details><summary><b>Expected output</b></summary>

```text
3
2
1
Liftoff!
```

</details>

<details><summary><b>Solution</b></summary>

```python
def countdown(n):
    if n == 0:
        print("Liftoff!")
        return
    print(n)
    countdown(n - 1)

countdown(3)
```

The base case is `n == 0`: print `Liftoff!` and `return`, so no further call is made. Every other call does one small piece of work (print `n`) and hands a *smaller* problem (`n - 1`) to the next call. Because each call moves closer to `0`, the base case is always reached.

</details>

<details><summary><b>Senior dev solution</b></summary>

```python
def countdown(n):
    for number in range(n, 0, -1):
        print(number)
    print("Liftoff!")

countdown(3)
```

A countdown is a straight line of work, and a loop says that directly. It also has no depth limit: `countdown(5000)` with recursion crashes (see Q27), while the loop is fine. Seniors save recursion for problems that branch or nest.

</details>

```python
prices = [4.5, 2.0, 3.5]
```

An empty list sums to `0`. A non-empty list sums to its first item plus the sum of everything after it (`numbers[1:]`).

```text
10.0
```

Each call peels off the first item and asks a recursive call for the sum of the rest. The list gets one shorter every time, so it eventually becomes `[]`, the base case. `4.5 + 2.0 + 3.5 + 0` is `10.0`.

### NB11-Q06 · L2 · Base Case · Error

This function has no base case. Which error does this raise?

**Which error does this raise?**

```python
def add_up(n):
    return n + add_up(n - 1)

print(add_up(3))
```

<details><summary><b>Hint</b></summary>

What stops the calls? Nothing tells `add_up` when to stop, so `n` keeps going: `2`, `1`, `0`, `-1`, ...

</details>

<details><summary><b>Expected output</b></summary>

```text
RecursionError
```

</details>

<details><summary><b>Solution</b></summary>

`RecursionError`: with no base case, every call makes another call. Each waiting call takes a slot on the call stack, and once there are about 1000 of them Python stops the program instead of running out of memory. When you see this error, check two things: is there a base case, and does every recursive call move *towards* it?

</details>

<details><summary><b>Senior dev solution</b></summary>

```python
def add_up(n):
    if n <= 0:
        return 0
    return n + add_up(n - 1)

print(add_up(3))
```

Seniors write the base case first, before the recursive call. Using `n <= 0` instead of `n == 0` also protects against a negative starting value, which would otherwise skip past `0` forever.

</details>

### NB11-Q07 · L2 · Call Stack · Predict

Code before and after the recursive call. Predict the output.

**What does this print?**

```python
def visit(n):
    if n == 0:
        print("bottom")
        return
    print("going down", n)
    visit(n - 1)
    print("coming back", n)

visit(2)
```

<details><summary><b>Hint</b></summary>

A call can't finish until the call it made has finished. Lines *after* `visit(n - 1)` wait until everything below has returned.

</details>

<details><summary><b>Expected output</b></summary>

```text
going down 2
going down 1
bottom
coming back 1
coming back 2
```

</details>

<details><summary><b>Solution</b></summary>

`visit(2)` prints `going down 2`, then calls `visit(1)` and *waits*. `visit(1)` prints `going down 1` and calls `visit(0)`, which prints `bottom` and returns. Now `visit(1)` resumes and prints `coming back 1`, then `visit(2)` resumes and prints `coming back 2`. The waiting calls are the call stack, and they finish in reverse order: last in, first out.

</details>

<details><summary><b>Senior dev solution</b></summary>

Work done *before* the recursive call happens on the way down; work done *after* it happens on the way back up, in reverse order. Seniors use this on purpose, e.g. printing a folder's name before its contents, or adding up sizes after the contents are known (see the mini-project).

Write a recursive function `is_palindrome(text)` that returns `True` if
`text` reads the same backwards. Don't reverse the string: compare the
first and last characters, then check what's between them. Print each
word and its result.

```text
racecar True
level True
python False
noon True
```

**Starting code (already in the cell below):**

```python
words = ["racecar", "level", "python", "noon"]
```

<details><summary><b>Hint</b></summary>

A string of 0 or 1 characters is a palindrome. If the first and last characters differ, it isn't. Otherwise it depends on `text[1:-1]`.

</details>

<details><summary><b>Expected output</b></summary>

```text
racecar True
level True
python False
noon True
```

</details>

<details><summary><b>Solution</b></summary>

```python
def is_palindrome(text):
    if len(text) <= 1:
        return True
    if text[0] != text[-1]:
        return False
    return is_palindrome(text[1:-1])

for word in words:
    print(word, is_palindrome(word))
```

This function has *two* base cases: a short string (`True`) and a mismatched pair (`False`). Only when the outer letters match does it recurse on the middle, `text[1:-1]`, which is two characters shorter each time. "noon" shrinks to "oo", then to `""`, which is `True`.

</details>

<details><summary><b>Senior dev solution</b></summary>

```python
def is_palindrome(text):
    cleaned = text.casefold()
    return cleaned == cleaned[::-1]

for word in words:
    print(word, is_palindrome(word))
```

Comparing with the reversed string is the direct way. `casefold()` (NB02) makes "Racecar" count too. Real palindrome checkers also strip spaces and punctuation, so that "Never odd or even" passes.

### NB11-Q10 · L2 · Base Case · Refactor

This recursive `length()` works, but it has far more base cases than it

```python
def length(items):
    if items == []:
        return 0
    elif len(items) == 1:
        return 1
    elif len(items) == 2:
        return 2

        count = 1 + length(items[1:])

print(length(["tea", "milk", "bread", "eggs"]), length([]))
```

<details><summary><b>Hint</b></summary>

Which base cases does the recursive case already handle by itself? Try following a 2-item list through the `else` branch.

</details>

<details><summary><b>Expected output</b></summary>

```text
4 0
```

</details>

<details><summary><b>Solution</b></summary>

```python
def length(items):
    if items == []:
        return 0
    return 1 + length(items[1:])

print(length(["tea", "milk", "bread", "eggs"]), length([]))
```

A 2-item list goes `1 + length([one item])` = `1 + 1 + length([])` = `2` without any special case. The extra base cases only repeat what the recursive case already does, and every extra branch is a place for a bug to hide. Pick the **smallest** input as the base case and let recursion do the rest.

</details>

<details><summary><b>Senior dev solution</b></summary>

```python
def length(items):
    if not items:
        return 0
    return 1 + length(items[1:])

print(length(["tea", "milk", "bread", "eggs"]), length([]))
```

`if not items:` uses truthiness (NB03): any empty list, tuple or string is falsy, so the function now works for strings too. And of course, real code calls `len()`.

</details>

````python
def length(items):
    if items == []:
        return 0
    elif len(items) == 1:
        return 1
    elif len(items) == 2:
        return 2
    else:
        count = 1 + length(items[1:])
        return count

print(length(["tea", "milk", "bread", "eggs"]), length([]))
````

### NB11-Q14 · L3 · Nested Data · Write

Write a recursive function `flatten(items)` that returns a flat list of
every number inside `nested`, however deeply it's nested. Print it.

```text
[1, 2, 3, 4, 5, 6, 7]
```

**Starting code (already in the cell below):**

```python
nested = [1, [2, 3], [4, [5, [6]]], 7]
```

<details><summary><b>Hint</b></summary>

Loop over the items. `isinstance(item, list)` tells you whether an item is itself a list that needs flattening.

</details>

<details><summary><b>Expected output</b></summary>

```text
[1, 2, 3, 4, 5, 6, 7]
```

</details>

<details><summary><b>Solution</b></summary>

```python
def flatten(items):
    flat = []
    for item in items:
        if isinstance(item, list):
            flat.extend(flatten(item))
        else:
            flat.append(item)
    return flat

print(flatten(nested))
```

A loop handles the *width* (each item at this level) and recursion handles the *depth* (lists inside lists). A plain number is added as it is. A list is flattened by a recursive call, and its flat result is added with `extend()`. No matter how deep the nesting goes, each call only has to deal with one level.

</details>

<details><summary><b>Senior dev solution</b></summary>

```python
def flatten(items):
    for item in items:
        if isinstance(item, list):
            yield from flatten(item)
        else:
            yield item

print(list(flatten(nested)))
```

A generator (NB22) hands out the numbers one at a time instead of building a new list at every level, which saves memory on large data. `yield from` passes along everything the inner call produces.

### NB11-Q15 · L3 · Divide and Conquer · Write

Write a recursive `binary_search(items, target, low=0, high=None)` that
returns the index of `target` in the sorted list, or `-1` if it isn't
there. Print the results for `31` and `10` on one line: `5 -1`.

Check the middle item, then search only the half that could hold the
target. Don't slice the list: pass `low` and `high` instead.

**Starting code (already in the cell below):**

```python
prices = [3, 8, 12, 19, 25, 31, 40, 52]
```

<details><summary><b>Hint</b></summary>

If `high` is `None`, start at the last index. If `low > high`, the range is empty: return `-1`. Otherwise compare `items[mid]` with the target and recurse on one half.

</details>

<details><summary><b>Expected output</b></summary>

```text
5 -1
```

</details>

<details><summary><b>Solution</b></summary>

```python
def binary_search(items, target, low=0, high=None):
    if high is None:
        high = len(items) - 1
    if low > high:
        return -1
    mid = (low + high) // 2
    if items[mid] == target:
        return mid
    if items[mid] < target:
        return binary_search(items, target, mid + 1, high)
    return binary_search(items, target, low, mid - 1)
```

<details><summary><b>Hint</b></summary>

</details>

<details><summary><b>Expected output</b></summary>


```python
def merge(left, right):
    result = []
    i = 0
    j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

def merge_sort(items):
    if len(items) <= 1:
        return items
    mid = len(items) // 2
    return merge(merge_sort(items[:mid]), merge_sort(items[mid:]))

print(merge_sort(scores))
```

### NB11-Q29 · L5 · Backtracking · Write

Place `n` queens on an `n` × `n` chessboard so no two attack each other
(no shared row, column or diagonal). Count the solutions for `n` = 4, 6
and 8 with backtracking:

```text
4: 2

8: 92
```

Place one queen per row. Keep a list `cols` where `cols[r]` is the column
of the queen in row `r`.

<details><summary><b>Hint</b></summary>

For the next row, try every column. It's safe if no earlier queen has the same column, and no earlier queen is on a diagonal: two queens share a diagonal when their column distance equals their row distance. Place, recurse, then remove.

</details>

<details><summary><b>Expected output</b></summary>

```text
4: 2
6: 4
8: 92
```

</details>

<details><summary><b>Solution</b></summary>

```python
def is_safe(cols, col):
    row = len(cols)
    for r in range(row):
        if cols[r] == col or abs(cols[r] - col) == row - r:
            return False
    return True

def count_queens(n, cols):
    if len(cols) == n:
        return 1
    total = 0
    for col in range(n):
        if is_safe(cols, col):
            cols.append(col)
            total += count_queens(n, cols)
            cols.pop()
    return total

for n in [4, 6, 8]:
    print(f"{n}: {count_queens(n, [])}")
```

Each call places a queen in the next row. If a column is safe, it places the queen, counts every way to finish the board from there, then *removes* the queen (`pop()`) to try the next column. That undo step is what makes it backtracking. A full board (`len(cols) == n`) is one solution. Two traps: the diagonal test needs `abs()` (diagonals go both ways), and forgetting `pop()` leaves old queens on the board.

</details>

<details><summary><b>Senior dev solution</b></summary>

```python
def count_queens(n):
    cols, downs, ups = set(), set(), set()

    def place(row):
        if row == n:
            return 1
        found = 0
        for col in range(n):
            if col in cols or row - col in downs or row + col in ups:
                continue
            cols.add(col)
            downs.add(row - col)
            ups.add(row + col)
            found += place(row + 1)
            cols.remove(col)
            downs.remove(row - col)
            ups.remove(row + col)
        return found

    return place(0)

for n in [4, 6, 8]:
    print(f"{n}: {count_queens(n)}")
```

Every square on a down-diagonal has the same `row - col`, and every square on an up-diagonal has the same `row + col`. Keeping three sets turns the safety check into three instant lookups instead of a loop over earlier rows, which matters as `n` grows.

```python
OPS = {"+": operator.add, "-": operator.sub, "*": operator.mul, "/": operator.truediv}

            return f"({show(left)} {op} {show(right)})"
        case _:

            return OPS[op](evaluate(left), evaluate(right))
        case _:
            return node
```

```python
def size(node):
    if not isinstance(node, dict):
        return node
    total = 0
    for child in node.values():
        total += size(child)
    return total

def show(name, node, level):
    indent = "  " * level
    if isinstance(node, dict):
        print(f"{indent}{name}/ ({size(node)} KB)")
        for child_name, child in node.items():
            show(child_name, child, level + 1)
    else:
        print(f"{indent}{name} ({node} KB)")

def largest(node):
    best_name = ""
    best_size = -1
    for child_name, child in node.items():
        if isinstance(child, dict):
            child_size = size(child)
            if child_size > best_size:
                best_name, best_size = child_name, child_size
            inner_name, inner_size = largest(child)
            if inner_size > best_size:
                best_name, best_size = inner_name, inner_size
    return best_name, best_size

show("home", home, 0)
print(f"Total: {size(home)} KB")
name, folder_size = largest(home)
print(f"Largest folder: {name} ({folder_size} KB)")
```

All three functions follow the same pattern: a file is the base case, a folder loops over its children and recurses. `show` prints *before* recursing (so a folder appears above its contents), while `size` adds *after* its recursive calls return (a folder's size is only known once its contents are). `largest` has to look inside every folder, because the biggest one might be nested deep down; starting `best_size` at `-1` means even an empty folder like `music/` can be compared. An empty dict also needs no special case: its loop simply doesn't run, giving `0 KB`.

</details>

<details><summary><b>Senior dev solution</b></summary>

```python
def walk(name, node, level, lines, folders):
    indent = "  " * level
    if not isinstance(node, dict):
        lines.append(f"{indent}{name} ({node} KB)")
        return node
    header = len(lines)
    lines.append("")
    total = sum(walk(child_name, child, level + 1, lines, folders)
                for child_name, child in node.items())
    lines[header] = f"{indent}{name}/ ({total} KB)"
    if level > 0:
        folders.append((total, name))
    return total

lines, folders = [], []
total = walk("home", home, 0, lines, folders)
print("\n".join(lines))
print(f"Total: {total} KB")
biggest_size, biggest_name = max(folders, key=lambda pair: pair[0])
print(f"Largest folder: {biggest_name} ({biggest_size} KB)")
```

The beginner solution calls `size()` again for every folder at every level, so deep trees get re-added many times.
