<details><summary><b>Solution</b></summary>

```python
def find(ids, target, low, high):
    if low > high:
        return -1
    mid = (low + high) // 2
    if ids[mid] == target:
        return mid
    if ids[mid] < target:
        return find(ids, target, mid + 1, high)
    return find(ids, target, low, mid - 1)
```

### Q15 · Build

The naive `fib(n)` calls itself twice and recomputes the same values again
and again, which is painfully slow for big `n`. Write `fib(n, memo)` that stores results in the dict `memo` and checks
it first. Print `fib(80)` starting with an empty dict.

**Expected output:**

```text
23416728348467685
```

<details><summary><b>Solution</b></summary>

```python
def fib(n, memo):
    if n in memo:
        return memo[n]
    if n < 2:
        return n
    memo[n] = fib(n - 1, memo) + fib(n - 2, memo)
    return memo[n]
print(fib(80, {}))
```

<details><summary><b>Solution</b></summary>

```python
def merge_sort(items):
    if len(items) <= 1:
        return items
    mid = len(items) // 2
    left = merge_sort(items[:mid])
    right = merge_sort(items[mid:])
    merged = []
    i = 0
    j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged.append(left[i])
            i += 1
        else:
            merged.append(right[j])
            j += 1
    return merged + left[i:] + right[j:]
print(merge_sort(scores))
```
