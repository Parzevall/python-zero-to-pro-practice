### Step 5

Write `report(sales)` that prints the grand total (compute it with `functools.reduce`), then each region with its share of the grand total as a percentage rounded to 1 decimal, biggest region first. Call it.

**Expected output:**

```text
Grand total: 3420

```

<details><summary><b>Solution</b></summary>

```python
from functools import reduce
def report(sales):
    grand = reduce(lambda total, s: total + s["qty"] * s["price"], sales, 0)
    print("Grand total:", grand)
    by_region = {}
    for s in sales:
        by_region[s["region"]] = by_region.get(s["region"], 0) + s["qty"] * s["price"]
    for region, total in sorted(by_region.items(), key=lambda pair: (-pair[1], pair[0])):
        print(f"{region}: {round(total / grand * 100, 1)}%")
report(sales)
```
