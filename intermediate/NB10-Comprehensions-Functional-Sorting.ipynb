# NB10 · Comprehensions, Functional Tools & Sorting

`intermediate/` · notebook 10 of 32 · 40 questions + a mini-project

Most loops you've written so far follow one of a few shapes: build a new
list from an old one, keep only some items, add things up, or find the
best one. This notebook gives each shape a shorter, clearer form.

- **Comprehensions** build a list, dict or set in one expression:
  `[price * 2 for price in prices if price > 5]`. Read it as "*this*, for
  each item, if *that*".
- **Generator expressions** look the same with round brackets. They hand
  out values one at a time instead of building a list, and they can only
  be used **once**.
- **`key=` functions** tell `sorted`, `min` and `max` *what to compare*.
  This is where the lambdas from NB09 earn their keep.

A comprehension is a tool, not a goal. If a loop has several steps, prints
things, or needs a comment to explain it, a plain `for` loop is often the
better choice. Several questions ask you to make that call.

### NB10-Q15 · L2 · Conditional expressions · Fix

```python
print([r if r > 0 else 0 for r in readings])
```

```python
readings = [5, -2, 3, -7]
print([max(r, 0) for r in readings])
```

"Clamp at zero" is exactly what `max(r, 0)` means, with no condition to read at all. Look for a built-in before writing a conditional.

</details>
