**Starting code (already in the cell below):**


<details><summary><b>Hint</b></summary>

An empty string counts as `False` in a condition, and any other string as `True`.

</details>

<details><summary><b>Expected output</b></summary>

```text

```

</details>

<details><summary><b>Solution</b></summary>

```python
if name:
    print(f"Hello, {name}!")
```

`if` doesn't need a `True`/`False` value: it asks whether the value is *truthy*. Empty strings, `0`, `0.0` and `None` are falsy; everything else is truthy. `""` is falsy, so the `else` block runs.

</details>

```python
print(f"Hello, {name or "guest"}!")
```

`name or "guest"` returns `name` if it's truthy, else `"guest"` (NB01). It's the senior idiom for defaults. Watch out: `"0"` is truthy but a count of `0` isn't, so use it only when every falsy value means "missing".

Use a `match` statement to print the meaning of an HTTP status code:

```python
match status:
```

`match` compares the value with each `case` in order and runs the first one that fits. `_` is the *wildcard*: it matches anything, so it goes last, like `else`. Unlike `switch` in other languages, there's no fall-through and no `break`.

</details>


The standard library already knows every status code (`http`, NB30). Seniors check for an existing table before writing one by hand. For your own codes, `match` is the right tool.

`command` arrives with stray spaces and in any case. Use `match` with `|`

```python
    case "pause":
        print("Paused")
    case unknown:
        print("Unknown command")
```

Matching on the cleaned expression keeps the raw input untouched for error messages. A bare name like `case unknown:` is a *capture pattern*: it matches anything and stores it in `unknown`, which you could print. Q28 shows the trap that comes with it.

`partition` gives `"7"` and `"05"`. Checking `isdigit()` first means `int()` can't fail, and an empty piece (no colon) fails `isdigit()` too. The trap is the greeting chain: starting with `5 <= hour < 12` and then `elif hour < 18` would call 3 a.m. the afternoon. Checking `hour < 5` first lets every later `elif` rely on a lower bound. `:02d` turns `7` into `07`.

</details>

<details><summary><b>Senior dev solution</b></summary>

```python
hour_text, _, minute_text = time_text.strip().partition(":")
valid = (hour_text.isdigit() and minute_text.isdigit() and len(minute_text) == 2
         and int(hour_text) <= 23 and int(minute_text) <= 59)
if not valid:
    print("Invalid time")
else:
    hour = int(hour_text)
    if 5 <= hour < 12:
        part = "morning"
    elif 12 <= hour < 18:
        part = "afternoon"
    elif 18 <= hour < 22:
        part = "evening"
    else:
        part = "night"
    print(f"Good {part}, it's {hour:02d}:{minute_text}")
```

`and` short-circuits, so `int()` only runs once `isdigit()` has passed: one `valid` flag replaces two branches. Writing *both* ends of each band (`12 <= hour < 18`) makes every branch correct on its own, whatever the order. For real times, `datetime.strptime` (NB15) does the parsing.

```python
if income < 0:
    print("Income can't be negative")
else:
    band_1 = min(income, 12000)
    band_2 = min(income, 50000) - 12000 if income > 12000 else 0
    band_3 = min(income, 150000) - 50000 if income > 50000 else 0
    band_4 = income - 150000 if income > 150000 else 0
    tax_2 = band_2 * 20 / 100
    tax_3 = band_3 * 40 / 100
    tax_4 = band_4 * 45 / 100
    total_tax = tax_2 + tax_3 + tax_4
    rate = total_tax / income if income > 0 else 0
    print(f"{"Income:":<16}{income:>10,.2f}")
    print(f"{"Band 1 (0%):":<16}{band_1:>10,.2f} taxed {0:>9,.2f}")
```
