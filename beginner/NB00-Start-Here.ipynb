Welcome. This notebook teaches you how to use the rest of the course, and
gets you comfortable with the one tool every Python program uses: `print()`.

**How a notebook runs.** Each grey box is a *cell*. Click into one and press
**Shift + Enter** to run it. Python remembers everything you've run so far,
in the order *you* ran it, not the order the cells sit on the page. If things
get strange, use **Kernel → Restart** and run the cells again from the top.

**Exploring on your own.** `help(print)` shows the documentation for
`print`, and `dir("hi")` lists everything a value can do. You won't need
them for these questions, but they're handy whenever you're curious.

**Reading an error.** When something goes wrong, Python prints a
*traceback*. Read the **last line** first: it names the error
(`TypeError`, `NameError`, ...) and says what went wrong.

**How your code becomes a result.** Python first reads your whole cell to
check the grammar. A grammar mistake is a `SyntaxError`, and *nothing* in
the cell runs. If the grammar is fine, it runs your lines top to bottom, and
a problem found while running stops it at that line.

### NB00-Q14 · L5 · print · Write

Use **four** `print()` calls to produce exactly this single line:

```text
Loading... 1 2 3 done
```

Rules: the first call prints `Loading...`, the next three print one number each, and the last word `done` comes from the fourth call.

<details><summary><b>Hint</b></summary>

Every call except the last needs `end=` so the next one continues on the same line. Think about where the spaces come from.

</details>

<details><summary><b>Expected output</b></summary>

```text
Loading... 1 2 3 done
```

</details>

<details><summary><b>Solution</b></summary>

```python
print("Loading...", end=" ")
print(1, end=" ")
print(2, end=" ")
print(3, "done")
```

`end=" "` replaces each line break with a space, so the output keeps going on one line. The last call prints `3`, then `print()`'s normal space separator, then `done`, then a normal line break.

</details>

<details><summary><b>Senior dev solution</b></summary>

```python
print("Loading...", 1, 2, 3, "done")
```

Without the four-call rule, one call does it, since `print()` already spaces its values. The four-call version mirrors real progress output, where each number is printed as a step finishes. With loops (NB04) that becomes `print(i, end=" ")` inside a `for`.

## NB00-P · L5 · Mini-project: Business Card · Write

Print this business card **exactly**, using only `print()` with `sep=` and `end=` where they help:

```text
+--------------------------+
|  Ada Lovelace            |
|  Analytical Engines Ltd  |
```

Count the spaces. Every line between the borders is the same width.

<details><summary><b>Hint</b></summary>

The top and bottom borders are identical. Two spaces after each `|`, then pad with spaces until the closing `|` lines up.

</details>

<details><summary><b>Expected output</b></summary>

```text
+--------------------------+
|  Ada Lovelace            |
|  Analytical Engines Ltd  |
|  ada@engines.example     |
+--------------------------+
```

</details>

<details><summary><b>Solution</b></summary>

```python

print("|  Ada Lovelace            |")
print("|  Analytical Engines Ltd  |")
```

One `print()` per line is the clearest way to get exact spacing. The border is 28 characters: a `+`, 26 dashes and a `+`.

</details>

<details><summary><b>Senior dev solution</b></summary>

```python
border = "+" + "-" * 26 + "+"
print(border)
print(f"|  {'Ada Lovelace':<24}|")
print(f"|  {'Analytical Engines Ltd':<24}|")
print(f"|  {'ada@engines.example':<24}|")
print(border)
```

A senior would never count spaces by hand. `"-" * 26` builds the border (NB01), and the f-string format `:<24` pads text to 24 characters on the left (NB02). Change the name, and the box still lines up.

</details>
