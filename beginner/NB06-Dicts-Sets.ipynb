Rewrite the loop so it gets each name and score together.

<details><summary><b>Hint</b></summary>

<details><summary><b>Solution</b></summary>

```python
scores = {"Ada": 91, "Grace": 85, "Alan": 78}
for name, score in scores.items():
```

Looping over a dict gives you its keys. `.items()` gives `(key, value)` pairs instead, and `name, score` unpacks each pair (like tuple unpacking in NB05). Dicts remember insertion order, so the names come out in the order they were added.

</details>

<details><summary><b>Senior dev solution</b></summary>

```python
scores = {"Ada": 91, "Grace": 85, "Alan": 78}
for name, score in scores.items():
    print(f"{name}: {score}")
```

`for key, value in d.items()` is the standard way to walk a dict. An f-string (NB02) builds the line in one place, so changing the format later means editing one string.

</details>

### NB06-Q38 · L5 · Sliding window · Write

<details><summary><b>Hint</b></summary>

Keep a window `start..i`. A dict remembers the last position of each character. When you meet a character already inside the window, move `start` just past its last position. Careful: a character seen *before* the window started doesn't count.

</details>

<details><summary><b>Expected output</b></summary>

```text
3 abc
2 ab
3 wke
```

</details>

<details><summary><b>Solution</b></summary>

```python
for text in texts:
    last_seen = {}
    start = 0
    best_start = 0
    best_len = 0
    for i, ch in enumerate(text):
        if ch in last_seen and last_seen[ch] >= start:
            start = last_seen[ch] + 1
        last_seen[ch] = i
        if i - start + 1 > best_len:
            best_len = i - start + 1
            best_start = start
    print(best_len, text[best_start:best_start + best_len])
```
