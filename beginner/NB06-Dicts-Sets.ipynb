## Q20 · Story project: Corner Shop Orders

You run the corner shop's order system. Today's orders came in as a list
of `(item, quantity)` pairs. Run them against the stock, then report.
Each step builds on the one before.

**Run these cells first.** They set up the starting files and data for every step below.

````python
stock = {"apple": 20, "batteries": 3, "bread": 5, "milk": 8, "rice": 7, "soap": 2, "tea": 12}
price = {"apple": 0.5, "batteries": 5.0, "bread": 2.0, "milk": 1.2, "rice": 1.8, "soap": 3.0, "tea": 4.0}
category = {"apple": "food", "batteries": "home", "bread": "food", "milk": "food", "rice": "food",
            "soap": "home", "tea": "drinks"}
orders = [("apple", 6), ("soap", 3), ("milk", 8), ("tea", 2), ("bread", 1), ("salt", 1), ("apple", 10)]
````

### Step 1

Run each order: if the item is in stock with enough quantity, subtract it from `stock` and add the order to a `filled` list; otherwise add the item name to a `shortages` list. Print both lists.

**Expected output:**

```text
Filled: [('apple', 6), ('milk', 8), ('tea', 2), ('bread', 1), ('apple', 10)]
Shortages: ['soap', 'salt']
```

<details><summary><b>Solution</b></summary>

```python
filled = []
shortages = []
for item, qty in orders:
    if item in stock and stock[item] >= qty:
        stock[item] -= qty
        filled.append((item, qty))
    else:
        shortages.append(item)
print("Filled:", filled)
print("Shortages:", shortages)
```

</details>
