### Q12 · Real-life

Write `password_ok(password)` that returns `True` only if the password
has at least 8 characters, contains a digit and contains an upper-case
letter. Use an early `return False` for each failed rule. Print the
verdict for each candidate.

**Expected output:**

```text
hunter2 False

SUNSHINEEE False
```

<details><summary><b>Solution</b></summary>

```python
def password_ok(password):
    if len(password) < 8:
        return False
    has_digit = False
    has_upper = False
    for ch in password:
        if ch.isdigit():
            has_digit = True
        if ch.isupper():
            has_upper = True
    return has_digit and has_upper
for candidate in candidates:
    print(candidate, password_ok(candidate))
```

### Q16 · Puzzle

Write `to_roman(n)` for 1 to 3999 using value/symbol pairs from largest
to smallest (1000 M, 900 CM, 500 D, 400 CD, 100 C, 90 XC, 50 L, 40 XL,
10 X, 9 IX, 5 V, 4 IV, 1 I). Print the Roman numeral for each year.

**Expected output:**

```text
1994 MCMXCIV
2024 MMXXIV
49 XLIX
```

<details><summary><b>Solution</b></summary>

```python
def to_roman(n):
    values = [1000, 900, 500, 400, 100, 90, 50, 40, 10, 9, 5, 4, 1]
    symbols = ["M", "CM", "D", "CD", "C", "XC", "L", "XL", "X", "IX", "V", "IV", "I"]
    result = ""
    for i in range(len(values)):
        while n >= values[i]:
            result += symbols[i]
            n -= values[i]
    return result
for year in years:
    print(year, to_roman(year))
```

### Q17 · Puzzle

Write `my_split(text, sep=" ")` that behaves like `text.split(sep)` for a

<details><summary><b>Solution</b></summary>

```python
def my_split(text, sep=" "):
    parts = []
    current = ""
    for ch in text:
        if ch == sep:
            parts.append(current)
            current = ""
        else:
            current += ch
    parts.append(current)
    return parts

print(my_split(line, ","))
```


<details><summary><b>Solution</b></summary>

```python
def top_words(words):
    counts = {}
    for word in words:
        counts[word] = counts.get(word, 0) + 1
    best = words[0]
    for word in counts:
        if counts[word] > counts[best]:
            best = word
    return best, counts[best]
word, times = top_words(get_words(text))
```
