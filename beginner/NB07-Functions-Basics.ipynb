Write a function `square(n)` that returns `n` multiplied by itself. Then


<details><summary><b>Hint</b></summary>

</details>


```python
def square(n):

```

`def` creates the function but doesn't run it. `square(7)` *calls* it: `n` becomes `7`, the body runs, and `return` hands `49` back to the caller, which passes it to `print()`.

```python
    return n ** 2

```

`n ** 2` says "squared" directly. The bigger senior habit is that the function *returns* and the caller decides what to do with the result (print it, store it, add it up).

### NB07-Q11 · L2 · Scope · Error

The function tries to add 1 to a global counter. Which error does this raise?

**Which error does this raise?**

```python
visits = 0

def record_visit():
    visits += 1

record_visit()
print(visits)
```

<details><summary><b>Hint</b></summary>

`visits += 1` *assigns* to `visits`. When a function assigns to a name, Python decides that name is local to the function.

</details>

<details><summary><b>Expected output</b></summary>

```text
UnboundLocalError
```

</details>

<details><summary><b>Solution</b></summary>

`UnboundLocalError`: because the function assigns to `visits`, Python treats `visits` as a *local* name for the whole function. `visits += 1` first needs to read the local `visits`, which doesn't have a value yet. The global `visits = 0` is never looked at. (It's a kind of `NameError`.)

</details>

<details><summary><b>Senior dev solution</b></summary>

```python
def record_visit(visits):
    return visits + 1

visits = 0
visits = record_visit(visits)
print(visits)
```

Q22 shows the `global` keyword that makes the original work. Seniors usually prefer this version instead: pass the value in, return the new value, and let the caller store it.

</details>

### NB07-Q16 · L2 · Pure functions · Write

Write `with_discount(prices, percent)` that returns a **new** list of
discounted prices and leaves the original list alone. Print the new list,
then the original.

```text
[9.0, 18.0, 4.5]
[10.0, 20.0, 5.0]
```

**Starting code (already in the cell below):**

```python
prices = [10.0, 20.0, 5.0]
```

<details><summary><b>Hint</b></summary>

Build a fresh list inside the function with `.append()`, and return it.

</details>

<details><summary><b>Expected output</b></summary>

```text
[9.0, 18.0, 4.5]
[10.0, 20.0, 5.0]
```

</details>

<details><summary><b>Solution</b></summary>

```python
def with_discount(prices, percent):
    result = []
    for price in prices:
        result.append(price * (1 - percent / 100))
    return result

print(with_discount(prices, 10))
print(prices)
```

The function reads `prices` but never changes it: it builds `result` and returns that. A function like this, whose output depends only on its inputs and which changes nothing outside itself, is called *pure*. Pure functions are easy to test and safe to call twice.

</details>

<details><summary><b>Senior dev solution</b></summary>

```python
def with_discount(prices, percent):
    factor = 1 - percent / 100
    return [price * factor for price in prices]

print(with_discount(prices, 10))
print(prices)
```

A list comprehension (NB10) builds the new list in one line, and the factor is worked out once instead of on every item. Seniors default to pure functions and make mutation the clearly named exception.

```python
                word += ch
```

```python
        counts[word] = counts.get(word, 0) + 1
```

```python
    return sum(map(len, words)) / len(words)

def top_words(text, n):
    """Return the n most common (word, count) pairs, ties alphabetical."""
    counts = Counter(get_words(text))
    return sorted(counts.items(), key=lambda pair: (-pair[1], pair[0]))[:n]

def longest_sentence(text):
    """Return (sentence, word count) for the sentence with the most words."""
    sentences = text.replace("!", ".").replace("?", ".").split(".")
    best = max(sentences, key=lambda s: len(s.split()))
    return best.strip(), len(best.split())

def report(text):
    """Print a short statistics report for text."""
    print("Words:", len(get_words(text)))
    print(f"Average word length: {average_word_length(text):.2f}")
    print("Top 3:", ", ".join(f"{w} ({c})" for w, c in top_words(text, 3)))
    sentence, count = longest_sentence(text)
    print(f"Longest sentence: {sentence} ({count} words)")

report(text)
```
