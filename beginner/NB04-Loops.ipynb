# NB04 · Loops

`beginner/` · notebook 4 of 32 · 38 questions + a mini-project

Loops let one piece of code run many times: once per letter, once per
number in a range, or for as long as a condition holds. Almost every real
program has one.

You already know variables, strings and `if`. That's all a loop needs.
Lists arrive in NB05, so here you loop over **strings**, over **`range()`**,
and over the pieces you get from `text.split()` (NB02). Where several
numbers are needed, they come packed in one string like `"6 9 8 7"`.

The traps in this notebook are the classics: `range` stopping one early, a
`while` loop that never ends, and the `else` clause that runs only when the
loop did **not** `break`.

```python
scores = "6 9 8 7"
```

<details><summary><b>Hint</b></summary>

How many times does the line `total = 0` run?

</details>

<details><summary><b>Expected output</b></summary>

</details>

<details><summary><b>Solution</b></summary>

```python
total = 0
count = 0
for text in scores.split():
    total += int(text)
    count += 1

```

`total = 0` was inside the loop, so it reset the total on every turn. By the end, `total` held only the last score, 7, and `7 / 4` is `1.75`. An accumulator must be set up *once*, before the loop starts.

```python
print(total / count if count else 0)
```

Seniors also think about the empty case: with `scores = ""`, `count` stays 0 and the division raises `ZeroDivisionError`. The conditional expression (NB03) decides what an empty average means. With lists (NB05) this becomes `sum(numbers) / len(numbers)`.

### NB04-Q19 · L3 · Accumulators · Write

`steps` holds a week of daily step counts, Monday first. Print the highest
count and which day number it was on (day 1 is Monday). If two days tie,
report the earlier one. Don't use `max()`.

```text
Best day: 4 with 12010 steps
```

**Starting code (already in the cell below):**

```python
steps = "8200 10450 7300 12010 9900 12010 6400"
```

<details><summary><b>Hint</b></summary>

Keep two "best so far" variables. Start the count at `None` so the first day always wins, and use `enumerate(..., start=1)` for day numbers. Think about `>` versus `>=` for the tie.

</details>

<details><summary><b>Expected output</b></summary>

```text
Best day: 4 with 12010 steps
```

</details>

<details><summary><b>Solution</b></summary>

```python
best_steps = None
best_day = 0
for day, text in enumerate(steps.split(), start=1):
    count = int(text)
    if best_steps is None or count > best_steps:
        best_steps = count
        best_day = day
print(f"Best day: {best_day} with {best_steps} steps")
```

This is the *max so far* pattern: compare each value with the best seen yet, and replace it only when the new one is bigger. Using `>` (not `>=`) means a tie doesn't replace the earlier day, so day 4 wins over day 6. Starting at `None` works for any numbers, even negative ones (see Q33).

</details>

<details><summary><b>Senior dev solution</b></summary>

```python
counts = [int(text) for text in steps.split()]
```

The Fibonacci sequence starts `1, 2` here, and each next number is the sum

```text

```

**Starting code (already in the cell below):**

```python
limit = 4_000_000
```

<details><summary><b>Hint</b></summary>

Keep the two latest numbers in `a` and `b`, and move forward with `a, b = b, a + b`. Check `a` against the limit *before* you use it.

</details>

<details><summary><b>Expected output</b></summary>

```text
Even terms: 11
Sum: 4613732
```

</details>

<details><summary><b>Solution</b></summary>

```python
a, b = 1, 2
count = 0
total = 0
while a <= limit:
    if a % 2 == 0:
        count += 1
        total += a
    a, b = b, a + b
print("Even terms:", count)
print("Sum:", total)
```

Multiple assignment works out `b, a + b` *before* changing either name, so no temporary variable is needed (the swap trick from NB01). Testing `a <= limit` at the top means a number past the limit is never added. "Does not exceed" means `<=`, not `<`.

</details>

## NB04-P · L5 · Mini-project: Guessing Game Replay · Write

A guessing game was played, and you have the log. Replay it: for each
guess in `guesses`, print whether it was too high, too low or correct.

Rules:

- A guess that isn't a whole number is skipped (`Skipped 'x': not a number`)
  and doesn't count as a guess.
- Stop at the correct guess. Later guesses are ignored.
- If no guess was correct, print `Not solved. The number was 41.` instead
  of the `Solved in ...` line.
- Last, print the closest wrong guess and how far off it was. On a tie, keep
  the earlier one. If there were no wrong guesses, print
  `Closest miss: none`.

```text
Guess 1: 50 is too high
Guess 2: 25 is too low
Skipped 'x': not a number
Guess 3: 37 is too low
Guess 4: 43 is too high
Guess 5: 41 is correct!
Solved in 5 guesses.
Closest miss: 43 (off by 2)
```

Once it works, try `guesses = "10 90 40"` and `guesses = "41"` to test the
other endings.

**Starting code (already in the cell below):**

```python
secret = 41
guesses = "50 25 x 37 43 41 12"
```

<details><summary><b>Hint</b></summary>

Loop over `guesses.split()`. Use `.isdigit()` (NB02) with `continue` for the skips, `break` on the right guess, and the loop's `else` for "not solved". Track the closest miss with the *best so far* pattern, starting from `None`.

</details>

<details><summary><b>Expected output</b></summary>

```text
Guess 1: 50 is too high
Guess 2: 25 is too low
Skipped 'x': not a number
Guess 3: 37 is too low
Guess 4: 43 is too high
Guess 5: 41 is correct!
Solved in 5 guesses.
Closest miss: 43 (off by 2)
```

</details>

<details><summary><b>Solution</b></summary>

```python
attempts = 0
closest = None
closest_gap = None
for text in guesses.split():
    if not text.isdigit():
        print(f"Skipped '{text}': not a number")
        continue
    guess = int(text)
    attempts += 1
    if guess == secret:
        print(f"Guess {attempts}: {guess} is correct!")
        print(f"Solved in {attempts} guesses.")
        break
    if guess > secret:
        print(f"Guess {attempts}: {guess} is too high")
    else:
        print(f"Guess {attempts}: {guess} is too low")
    gap = abs(guess - secret)
    if closest_gap is None or gap < closest_gap:
        closest = guess
        closest_gap = gap
else:
    print(f"Not solved. The number was {secret}.")
if closest is None:
    print("Closest miss: none")
else:
    print(f"Closest miss: {closest} (off by {closest_gap})")
```

`continue` skips the bad entry *before* `attempts` goes up, so it doesn't count. `break` ends the game at the right guess, which also skips the loop's `else`; the `else` only runs when every guess was used up without a win. The closest miss is a *best so far* accumulator: `None` means "no miss yet", and `<` (not `<=`) keeps the earlier guess on a tie. The final `if` sits after the loop, so it runs for every ending.

</details>

<details><summary><b>Senior dev solution</b></summary>

```python

    if not text.isdigit():
        print(f"Skipped {text!r}: not a number")
        continue

    attempts += 1
    if guess == secret:
        print(f"Guess {attempts}: {guess} is correct!")
        noun = "guess" if attempts == 1 else "guesses"
        print(f"Solved in {attempts} {noun}.")
        break
    verdict = "too high" if guess > secret else "too low"
    print(f"Guess {attempts}: {guess} is {verdict}")
    if closest is None or abs(guess - secret) < abs(closest - secret):
        closest = guess
else:

    print("Closest miss: none")

```

Keeping only `closest` (and working out the gap from it) means two variables can never disagree.
