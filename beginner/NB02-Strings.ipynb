# NB02 · Strings

`beginner/` · notebook 2 of 32 · 41 questions + a mini-project

Text is everywhere: names, emails, file paths, log lines, receipts. In
Python, text is a `str`, and this notebook is about reading it, cutting it
up, cleaning it and printing it exactly the way you want.

You know variables, types, casting and operators from NB01. There are still
no `if` statements (NB03), loops (NB04) or lists (NB05). `split()` hands
you a list, and that's fine: here you only pass it straight to `join()`.
Lists get proper coverage in NB05.

Two ideas carry the whole notebook. First, a string is a sequence of
characters, so you can index it (`s[0]`) and slice it (`s[1:4]`). Second,
strings are **immutable**: methods like `.strip()` never change the string,
they hand you a new one.

<details><summary><b>Hint</b></summary>

Clean the names first (`strip`, then `title` or `lower`). The year's last two digits are a slice. `partition("@")` splits the email. For the card, `f"{label:<6}"` pads a label, and `f"{text:-^28}"` centres text in dashes.
