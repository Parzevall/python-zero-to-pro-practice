A web page shows `per_page` items. Print how many pages are needed for

`-(-a // b)` flips the sign, floors (which now rounds *up* in effect), and flips back. It's a common idiom, but `math.ceil(a / b)` is clearer, and seniors pick clear unless they need exact big-integer maths.
