it can still make the same trip back.

**Expected output:**

```text
Distance: 13.0
Battery left: 94.8
Can fly back: True
```

<details><summary><b>Solution</b></summary>

```python

dx = bx - ax
dy = by - ay
distance = math.sqrt(dx ** 2 + dy ** 2)
battery = 100 - distance * drain_per_unit
print("Distance:", distance)
print("Battery left:", round(battery, 1))
print("Can fly back:", battery >= distance * drain_per_unit)
```

</details>

````python
ax, ay = 2, 3
bx, by = 14, 8
drain_per_unit = 0.4

# your code here
````
